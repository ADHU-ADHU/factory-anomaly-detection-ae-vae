```python
input_shape = (400,320, 3)
datagen = ImageDataGenerator(rescale=1./255)
    
path_train = "/path/to/private-data/Datagen_Left/Outlet_Left/train"
path_val = "/path/to/private-data/Datagen_Left/Outlet_Left/validation"
    
batch_size = 32
target_size = (input_shape[0],input_shape[1])
train_generator = datagen.flow_from_directory(path_train, target_size=target_size,  batch_size=batch_size, class_mode='input', shuffle=True, color_mode='rgb')

validation_generator = datagen.flow_from_directory(path_val, target_size=target_size , batch_size=batch_size, class_mode='input', shuffle=True, color_mode='rgb')
```

```python


# Assuming 'input_shape' is defined elsewhere in your code.
input_img = Input(shape=input_shape) 

x1 =Cropping2D(cropping=((0, input_shape[0]//2), (0, input_shape[1]//2)))(input_img)
x1 = Conv2D(128, (3, 3), activation='relu', padding='same')(x1)
x1 = MaxPooling2D((2, 2), padding='same')(x1)
x1 = Conv2D(64, (3, 3), activation='relu', padding='same')(x1)
x1 = MaxPooling2D((2, 2), padding='same')(x1)
x1 = Conv2D(64, (3, 3), activation='relu', padding='same')(x1)
x1 = MaxPooling2D((2, 2), padding='same')(x1)
x1 = Conv2D(64, (3, 3), activation='relu', padding='same')(x1)
x1 = MaxPooling2D((2, 2), padding='same')(x1)
x1 = Conv2D(32, (3, 3), activation='relu', padding='same')(x1)
x1 = MaxPooling2D((2, 2), padding='same')(x1)
x1 = Conv2D(32, (3, 3), activation='relu', padding='same')(x1)
x1_end = MaxPooling2D((2, 2), padding='same', name='latent1')(x1)
model1 = Model(inputs=input_img, outputs=x1_end)
model1.summary()


x2 = Cropping2D(cropping=((0, input_shape[0]//2), (input_shape[1]//2, 0)))(input_img)
x2 = Conv2D(128, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2), padding='same')(x2)
x2 = Conv2D(64, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2), padding='same')(x2)
x2 = Conv2D(64, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2), padding='same')(x2)
x2 = Conv2D(64, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2), padding='same')(x2)
x2 = Conv2D(32, (3, 3), activation='relu', padding='same')(x2)
x2 = MaxPooling2D((2, 2), padding='same')(x2)
x2 = Conv2D(32, (3, 3), activation='relu', padding='same')(x2)
x2_end = MaxPooling2D((2, 2), padding='same',name='latent2')(x2)
model2 = Model(inputs=input_img, outputs=x2_end)
model2.summary()


x3 = Cropping2D(cropping=((input_shape[0]//2, 0), (0, input_shape[1]//2)))(input_img)
x3 = Conv2D(128, (3, 3), activation='relu', padding='same')(x3)
x3 = MaxPooling2D((2, 2), padding='same')(x3)
x3 = Conv2D(64, (3, 3), activation='relu', padding='same')(x3)
x3 = MaxPooling2D((2, 2), padding='same')(x3)
x3 = Conv2D(64, (3, 3), activation='relu', padding='same')(x3)
x3 = MaxPooling2D((2, 2), padding='same')(x3)
x3 = Conv2D(64, (3, 3), activation='relu', padding='same')(x3)
x3 = MaxPooling2D((2, 2), padding='same')(x3)
x3 = Conv2D(32, (3, 3), activation='relu', padding='same')(x3)
x3 = MaxPooling2D((2, 2), padding='same')(x3)
x3 = Conv2D(32, (3, 3), activation='relu', padding='same')(x3)
x3_end = MaxPooling2D((2, 2), padding='same',name='latent3')(x3)
model3 = Model(inputs=input_img, outputs=x3_end)
model3.summary()


x4 =Cropping2D(cropping=((input_shape[0]//2, 0), (input_shape[1]//2, 0)))(input_img)
x4 = Conv2D(128, (3, 3), activation='relu', padding='same')(x4)
x4 = MaxPooling2D((2, 2), padding='same')(x4)
x4 = Conv2D(64, (3, 3), activation='relu', padding='same')(x4)
x4 = MaxPooling2D((2, 2), padding='same')(x4)
x4 = Conv2D(64, (3, 3), activation='relu', padding='same')(x4)
x4 = MaxPooling2D((2, 2), padding='same')(x4)
x4 = Conv2D(64, (3, 3), activation='relu', padding='same')(x4)
x4 = MaxPooling2D((2, 2), padding='same')(x4)
x4 = Conv2D(32, (3, 3), activation='relu', padding='same')(x4)
x4 = MaxPooling2D((2, 2), padding='same')(x4)
x4 = Conv2D(32, (3, 3), activation='relu', padding='same')(x4)
x4_end = MaxPooling2D((2, 2), padding='same',name='latent4')(x4)
model4 = Model(inputs=input_img, outputs=x4_end)
model4.summary()

x1_flat = Flatten()(x1)
x2_flat = Flatten()(x2)
x3_flat = Flatten()(x3)
x4_flat = Flatten()(x4)

# Concatenate all flattened outputs
from tensorflow.keras.layers import Concatenate, Lambda, Reshape
from tensorflow.keras.layers import concatenate
latent_space = concatenate([x1_flat, x2_flat, x3_flat, x4_flat])
```

```python

# Decoder layers
x1 = Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1_end)
x1 = Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1)
x1 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1)
x1 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1)
x1 = Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1)
x1 = Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same')(x1)
x1 = Conv2D(3, (3, 3), activation='sigmoid', padding='same')(x1)
x1_image = Cropping2D(cropping=((28,28), (16,16)))(x1)
decoder1 = Model(inputs=x1_end, outputs=x1_image)
decoder1.summary()

x2 = Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2_end)
x2 = Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2)
x2 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2)
x2 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2)
x2 = Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2)
x2 = Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same')(x2)
x2 = Conv2D(3, (3, 3), activation='sigmoid', padding='same')(x2)
x2_image = Cropping2D(cropping=((28,28), (16,16)))(x2)
decoder2 = Model(inputs=x2_end, outputs=x2_image)


x3 = Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3_end)
x3 = Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3)
x3 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3)
x3 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3)
x3 = Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3)
x3 = Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same')(x3)
x3 = Conv2D(3, (3, 3), activation='sigmoid', padding='same')(x3)
x3_image = Cropping2D(cropping=((28, 28), (16, 16)))(x3)  # Adjust cropping as needed for alignment
decoder3 = Model(inputs=x3_end, outputs=x3_image)
decoder3.summary()



x4 = Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4_end)
x4 = Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4)
x4 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4)
x4 = Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4)
x4 = Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4)
x4 = Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same')(x4)
x4 = Conv2D(3, (3, 3), activation='sigmoid', padding='same')(x4)
x4_image = Cropping2D(cropping=((28, 28), (16, 16)))(x4)  # Adjust cropping as needed for alignment
decoder4 = Model(inputs=x4_end, outputs=x4_image)
decoder4.summary()



top_half = Concatenate(axis=2)([x1_image, x2_image])  # axis=2 is the width
bottom_half = Concatenate(axis=2)([x3_image, x4_image])

# Concatenate the top and bottom halves along the height
full_image = Concatenate(axis=1)([top_half, bottom_half])

```

```python
final_model = Model(inputs=input_img, outputs=full_image)
final_model.summary()
```

```python
def ssim_loss(y_true, y_pred):
    # Return the negative SSIM. We negate since a higher SSIM value should correspond to better performance, 
    # but we need a loss that should be minimized
    return 1 - tf.reduce_mean(tf.image.ssim(y_true, y_pred, max_val=1.0))
```

```python
import glob, os
 
in3_mse = []
in5_mse = []
outlet_mse = []

in3_cos = []
in5_cos = []
outlet_cos = []


in3_ssim = []
in5_ssim = []
outlet_ssim = []



path = "/path/to/private-data/Croped_Pictures/train/Left/*"
for directory_path in glob.glob(path):
    label=directory_path.split("\\")[-1]
    for image_path in glob.glob(os.path.join(directory_path,"*.bmp")):
        img = load_img(image_path)
        img = np.expand_dims(img,axis=0)
        img = img/255.0
        prediction = model.predict(img)      
        mse = np.mean((img - prediction) ** 2)
        loss_ssim = ssim_loss(img,prediction).numpy()
        
        vector1 = img.squeeze()
        vector2 = prediction.squeeze()
        vector1 = vector1.flatten()
        vector2 = vector2.flatten()
        vector1 = vector1 / np.linalg.norm(vector1)
        vector2 = vector2 / np.linalg.norm(vector2)
        cosine_similarity = np.dot(vector1, vector2)

        
        if label == "Inlet (3 marking)":
            in3_mse.append(mse)
            in3_cos.append(cosine_similarity)
            in3_ssim.append(loss_ssim)
            
        elif label == "Inlet (5 marking)":
            in5_mse.append(mse)
            in5_cos.append(cosine_similarity)
            in5_ssim.append(loss_ssim)
            
        elif label == "Outlet":
            outlet_mse.append(mse)
            outlet_cos.append(cosine_similarity)
            outlet_ssim.append(loss_ssim)
```
