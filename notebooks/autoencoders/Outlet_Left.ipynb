```python
from keras.models import Sequential
import tensorflow as tf
from keras.layers import Conv2D, MaxPooling2D, Dropout, Flatten, Dense, Conv2DTranspose, Cropping2D, UpSampling2D, Reshape

from tensorflow.keras.preprocessing.image import ImageDataGenerator

from tensorflow.keras.applications.vgg16 import VGG16
from tensorflow.keras.applications.xception import Xception
from tensorflow.keras.applications.inception_v3 import InceptionV3
from tensorflow.keras.models import Model
```

```python
input_shape = (400,320, 3)
```

```python
datagen = ImageDataGenerator(rescale=1./255)
    
path_train = "/path/to/private-data/Datagen_Left/Outlet_Left/train"
path_val = "/path/to/private-data/Datagen_Left/Outlet_Left/validation"
    
batch_size = 32
target_size = (input_shape[0],input_shape[1])
train_generator = datagen.flow_from_directory(path_train, target_size=target_size,  batch_size=batch_size, class_mode='input', shuffle=True, color_mode='rgb')

validation_generator = datagen.flow_from_directory(path_val, target_size=target_size , batch_size=batch_size, class_mode='input', shuffle=True, color_mode='rgb')
```

```python
model = Sequential()
# Encoder
model.add(Conv2D(128, (3, 3), activation='relu', padding='same', input_shape=input_shape))
model.add(MaxPooling2D((2, 2), padding='same'))
model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D((2, 2), padding='same'))
model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D((2, 2), padding='same'))
model.add(Conv2D(64, (3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D((2, 2), padding='same'))
model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D((2, 2), padding='same'))
model.add(Conv2D(32, (3, 3), activation='relu', padding='same'))
model.add(MaxPooling2D((2, 2), padding='same'))


model.add(Flatten())
model.add(Dense(512, activation='relu'))

model.add(Dense(8*10*64, activation='relu'))
#model.save('Latent_space_(Outlet_left).keras')

target_shape = (8, 10, 64)  # Example shape
model.add(Reshape(target_shape))


# Decoder
model.add(Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2D(3, (3, 3), activation='sigmoid', padding='same'))
model.add(Cropping2D(cropping=((56,56), (160,160))))

model.compile(loss='mean_squared_error', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
epochs = 50
steps_per_epoch = len(train_generator)
model.fit(train_generator, epochs=epochs, validation_data=validation_generator, batch_size=batch_size, steps_per_epoch=steps_per_epoch)
```

```python
from tensorflow.keras.models import load_model
#model = load_model('full_model(Outlet_left).keras')

def ssim_loss(y_true, y_pred):
    # Return the negative SSIM. We negate since a higher SSIM value should correspond to better performance, 
    # but we need a loss that should be minimized
    return 1 - tf.reduce_mean(tf.image.ssim(y_true, tf.cast(y_pred,tf.float64), max_val=1.0))
```

```python

errors = np.concatenate([val_outlet_mse,val_in3_mse,val_in5_mse,val_others_mse])
labels = np.concatenate([np.zeros(len(val_outlet_mse)), np.ones(len(val_in3_mse)+len(val_in5_mse)+len(val_others_mse))])
```

```python
from sklearn.metrics import roc_auc_score

auc_roc = roc_auc_score(labels, errors)
print(f"AUC-ROC: {auc_roc}")
```

```python
import numpy as np
from sklearn.metrics import roc_curve, precision_recall_curve, f1_score
from scipy.optimize import brentq
from scipy.interpolate import interp1d



fpr, tpr, thresholds = roc_curve(labels, errors)
eer = brentq(lambda x: 1. - x - interp1d(fpr, tpr)(x), 0., 1.)
eer_threshold = interp1d(fpr, thresholds)(eer)

precision, recall, thresholds_pr = precision_recall_curve(labels, errors)
fscores = 2 * (precision * recall) / (precision + recall)
idx = np.nanargmax(fscores)  # handle NaNs if any
best_f1_score = fscores[idx]
f1_threshold = thresholds_pr[idx]

best_precision_idx = np.argmax(precision)
best_recall_idx = np.argmax(recall)
precision_threshold = thresholds_pr[best_precision_idx-1]  # adjusting index as it's one less
recall_threshold = thresholds_pr[best_recall_idx-1]       # adjusting index as it's one less


print(f"Equal Error Rate (EER): {eer:.5f}, Threshold: {eer_threshold}")
print(f"Best F1 Score: {best_f1_score:.5f}, F1 Threshold: {f1_threshold}")
print(f"Best Precision: {precision[best_precision_idx]:.5f}, Precision Threshold: {precision_threshold}")
print(f"Best Recall: {recall[best_recall_idx]:.5f}, Recall Threshold: {recall_threshold}")

```
