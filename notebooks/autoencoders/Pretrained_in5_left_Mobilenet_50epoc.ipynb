```python


# Define input shape
input_shape = (400, 320, 3)  # This is the default input shape for ResNet50

# Load ResNet50 as the base model
base_model = MobileNetV2(include_top=False, weights='imagenet', input_shape=input_shape)

# Freeze the base_model
base_model.trainable = False

# Encoder - using ResNet-50
model = Sequential()
model.add(Input(shape=input_shape))
model.add(base_model)
model.add(Flatten())
model.add(Dense(512, activation='relu'))

model.add(Dense(8*10*64, activation='relu'))


target_shape = (8, 10, 64)  # Example shape
model.add(Reshape(target_shape))
# Decoder
# Add decoder layers here
model.add(Conv2DTranspose(256, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(128, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(64, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(32, (3, 3), strides=(2, 2), activation='relu', padding='same'))
model.add(Conv2DTranspose(16, (3, 3), strides=(2, 2), activation='relu', padding='same'))

model.add(Conv2D(3, (3, 3), activation='sigmoid', padding='same'))
model.add(Cropping2D(cropping=((56,56), (160,160))))

model.summary()


model.compile(loss='mean_squared_error', optimizer='adam', metrics=['accuracy'])
model.summary()
```

```python
#model.save('Pretrained_in5_leftInceptionResNetV2.keras')
def ssim_loss(y_true, y_pred):
    # Return the negative SSIM. We negate since a higher SSIM value should correspond to better performance, 
    # but we need a loss that should be minimized
    return 1 - tf.reduce_mean(tf.image.ssim(y_true, tf.cast(y_pred,tf.float64), max_val=1.0))
```

```python
import glob, os
 
in3_mse = []
in5_mse = []
outlet_mse = []

in3_cos = []
in5_cos = []
outlet_cos = []


in3_ssim = []
in5_ssim = []
outlet_ssim = []



path = "/path/to/private-data/Croped_Pictures/train/Left/*"
for directory_path in glob.glob(path):
    label=directory_path.split("\\")[-1]
    for image_path in glob.glob(os.path.join(directory_path,"*.bmp")):
        img = load_img(image_path)
        img = np.expand_dims(img,axis=0)
        img = img/255.0
        prediction = model.predict(img)      
        mse = np.mean((img - prediction) ** 2)
        loss_ssim = ssim_loss(img,prediction).numpy()
        
        vector1 = img.squeeze()
        vector2 = prediction.squeeze()
        vector1 = vector1.flatten()
        vector2 = vector2.flatten()
        vector1 = vector1 / np.linalg.norm(vector1)
        vector2 = vector2 / np.linalg.norm(vector2)
        cosine_similarity = np.dot(vector1, vector2)

        
        if label == "Inlet (3 marking)":
            in3_mse.append(mse)
            in3_cos.append(cosine_similarity)
            in3_ssim.append(loss_ssim)
            
        elif label == "Inlet (5 marking)":
            in5_mse.append(mse)
            in5_cos.append(cosine_similarity)
            in5_ssim.append(loss_ssim)
            
        elif label == "Outlet":
            outlet_mse.append(mse)
            outlet_cos.append(cosine_similarity)
            outlet_ssim.append(loss_ssim)
```

```python
import numpy as np
from sklearn.metrics import roc_curve, precision_recall_curve, f1_score
from scipy.optimize import brentq
from scipy.interpolate import interp1d



fpr, tpr, thresholds = roc_curve(labels, errors)
eer = brentq(lambda x: 1. - x - interp1d(fpr, tpr)(x), 0., 1.)
eer_threshold = interp1d(fpr, thresholds)(eer)

precision, recall, thresholds_pr = precision_recall_curve(labels, errors)
fscores = 2 * (precision * recall) / (precision + recall)
idx = np.nanargmax(fscores)  # handle NaNs if any
best_f1_score = fscores[idx]
f1_threshold = thresholds_pr[idx]

best_precision_idx = np.argmax(precision)
best_recall_idx = np.argmax(recall)
precision_threshold = thresholds_pr[best_precision_idx-1]  # adjusting index as it's one less
recall_threshold = thresholds_pr[best_recall_idx-1]       # adjusting index as it's one less


print(f"Equal Error Rate (EER): {eer:.5f}, Threshold: {eer_threshold}")
print(f"Best F1 Score: {best_f1_score:.5f}, F1 Threshold: {f1_threshold}")
print(f"Best Precision: {precision[best_precision_idx]:.5f}, Precision Threshold: {precision_threshold}")
print(f"Best Recall: {recall[best_recall_idx]:.5f}, Recall Threshold: {recall_threshold}")

```
